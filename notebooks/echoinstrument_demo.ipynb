# `EchoInstrument` - Demonstration

This notebook demonstrates how to use the `EchoInstrument` class from `echosims.utils`.

The class represents the ECHO instrument and its frequency channels.

## Setup

The cell below adds the repo root to `sys.path` so this notebook can import `echosims` without installing the package. This will become unnecessary once the project is pip-installable .

```python
import sys
from pathlib import Path
sys.path.insert(0, str(Path.cwd().parent))
```

## 1. Importing the class

Constructing the instrument requires no arguments. The channel configuration is loaded automatically from the YAML file.

```python
from echosims.utils import EchoInstrument

instrument = EchoInstrument()
```

## 2. List All Available Frequency Channels

`get_frequencies()` returns the list of all available frequency channels, in $\mathrm{GHz}$.

```python
frequencies = instrument.get_frequencies()
print(f"Number of channels: {len(frequencies)}")
print(f"Channels (GHz): {frequencies}")
```

## 3. Get the Channel Specifications for a Specific Frequency

`get_channel(frequency)` returns a dictionary containing the channel specifications for the specified frequency.

```python
frequency = 165
channel_specs = instrument.get_channel(frequency)
print(f"Channel Specifications for {frequency} GHz channel: {channel_specs}")
print(f'Type of the returned object: {type(channel_specs)}')
```

## 4. Get the Beam FWHM for a Specific Frequency Channel

`get_beam_fwhm(frequency)` returns the beam full width at half maximum (FWHM), in $\mathrm{arcmin}$, for the specified frequency channel.

```python
beam_fwhm = instrument.get_beam_fwhm(frequency)
print(f"Beam FWHM for the {frequency} GHz channel: {beam_fwhm} arcmin")
```

## 5. Get the Polarization Sensitivity for a Specific Frequency Channel

`get_polarization_sensitivity(frequency)` returns the polarization sensitivity, in $\mu\mathrm{K}\,\mathrm{arcmin}$, for the specified frequency channel.

```python
pol_sens = instrument.get_polarization_sensitivity(frequency)

print(f"Polarization sensitivity for the {frequency} GHz channel: {pol_sens} uK·arcmin")
```
